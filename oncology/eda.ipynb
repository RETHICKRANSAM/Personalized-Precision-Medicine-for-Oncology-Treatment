```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
file_path = r"C:\Users\venka\Downloads\Oncology_Precision_Medicine_Cleaned_100_Rows_With_Additional_Clinical_Data.csv3.xls"

df = pd.read_csv(file_path)

print(df.shape)
df.head()
```

```python
print("\n" + "=" * 60)
print("DATA INFORMATION")
print("=" * 60)

print(df.info())

print("\nStatistical Summary:")
print(df.describe(include="all"))
```

```python
print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

missing_values = df.isnull().sum()

print(missing_values)

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
print("\n" + "=" * 60)
print("DUPLICATE RECORDS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
print("\n" + "=" * 60)
print("CATEGORICAL VALUE CHECK")
print("=" * 60)

categorical_columns = df.select_dtypes(include=["object"]).columns

for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts(dropna=False))
```

```python
print("\n" + "=" * 60)
print("TOXICITY LEVEL DISTRIBUTION")
print("=" * 60)

print(df["Toxicity_Level"].value_counts())

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Toxicity_Level"
)

plt.title("Toxicity Level Distribution")
plt.xlabel("Toxicity Level")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("BIOMARKER VALUE VS TOXICITY")
print("=" * 60)

print(
    df.groupby("Toxicity_Level")["Biomarker_Value"]
      .agg(["count", "mean", "median", "min", "max"])
)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Toxicity_Level",
    y="Biomarker_Value"
)

plt.title("Biomarker Value vs Toxicity Level")
plt.xlabel("Toxicity Level")
plt.ylabel("Biomarker Value")

plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("DOSE VS TOXICITY")
print("=" * 60)

print(
    df.groupby("Toxicity_Level")["Dose_mg"]
      .agg(["count", "mean", "median", "min", "max"])
)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Toxicity_Level",
    y="Dose_mg"
)

plt.title("Treatment Dose vs Toxicity Level")
plt.xlabel("Toxicity Level")
plt.ylabel("Dose (mg)")

plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("GENE MUTATION VS TOXICITY")
print("=" * 60)

mutation_table = pd.crosstab(
    df["Gene_Mutation"],
    df["Toxicity_Level"]
)

print(mutation_table)

mutation_table.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Gene Mutation vs Toxicity Level")
plt.xlabel("Gene Mutation")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("TUMOR SIZE VS TOXICITY")
print("=" * 60)

print(
    df.groupby("Toxicity_Level")["Tumor_Size_mm"]
      .agg(["count", "mean", "median", "min", "max"])
)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Toxicity_Level",
    y="Tumor_Size_mm"
)

plt.title("Tumor Size vs Toxicity Level")
plt.xlabel("Toxicity Level")
plt.ylabel("Tumor Size (mm)")

plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("ECOG PERFORMANCE STATUS VS TOXICITY")
print("=" * 60)

ecog_table = pd.crosstab(
    df["ECOG_Performance_Status"],
    df["Toxicity_Level"]
)

print(ecog_table)

ecog_table.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("ECOG Performance Status vs Toxicity")
plt.xlabel("ECOG Performance Status")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("TREATMENT TYPE VS TOXICITY")
print("=" * 60)

treatment_table = pd.crosstab(
    df["Treatment_Type"],
    df["Toxicity_Level"]
)

print(treatment_table)

treatment_table.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)

plt.title("Treatment Type vs Toxicity Level")
plt.xlabel("Treatment Type")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("TREATMENT RESPONSE VS TOXICITY")
print("=" * 60)

response_table = pd.crosstab(
    df["Treatment_Response"],
    df["Toxicity_Level"]
)

print(response_table)

response_table.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Treatment Response vs Toxicity Level")
plt.xlabel("Treatment Response")
plt.ylabel("Number of Patients")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)

numeric_df = df.select_dtypes(include=["number"])

correlation = numeric_df.corr()

print(correlation.round(2))
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Numerical Feature Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
print("\n" + "=" * 60)
print("OUTLIER CHECK")
print("=" * 60)

numeric_columns = df.select_dtypes(include=["number"]).columns

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(
        f"{column}: {len(outliers)} possible outliers"
    )
```

```python
for column in numeric_columns:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=df,
        x=column,
        kde=True
    )

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()
```
